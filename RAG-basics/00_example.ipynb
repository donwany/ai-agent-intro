## Cosine Similarity

```python
import numpy as np

A = np.array([1, 2, 3])
B = np.array([4, 5, 6])

cosine_similarity = np.dot(A, B) / (
    np.linalg.norm(A) * np.linalg.norm(B)
)

print(cosine_similarity)
```
